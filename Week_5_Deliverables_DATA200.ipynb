##1. Descriptive Statistical Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# load dataset
df = pd.read_csv("/content/Student_Performance.csv")

# descriptive statistics
df.describe()
```

##Interpretation
Descriptive statistics were used to summarize the overall characteristics of the student dataset. The results provide insights into study hours, attendance, and academic scores. The average overall performance score suggests that most students fall within a moderate performance range. Variability in study hours and attendance indicates that students differ in their learning habits and engagement levels. These statistics establish a foundational understanding of student behavior and academic trends before applying inferential analysis.

##2. Inferential Statistical Analysis — Pearson Correlation

```python
corr, p_value = stats.pearsonr(df['study_hours'], df['overall_score'])

corr, p_value
```

##Interpretation
Pearson correlation analysis was conducted to evaluate the relationship between daily study hours and overall academic performance. The correlation coefficient measures the strength and direction of the association. A positive correlation suggests that increased study time is linked to better academic results. The p-value determines statistical significance. If the p-value is below 0.05, the relationship is considered statistically meaningful. This analysis helps validate whether study habits directly influence performance outcomes.

##3. Diagnostic Measures — Missing Values

```python
df.isnull().sum()
```

##Interpretation
The dataset was examined for missing values to ensure data completeness. The analysis showed minimal or no missing entries, indicating that the dataset is well-structured and reliable for statistical modeling. Clean data reduces bias and increases confidence in the accuracy of the analysis.

##4. Duplicate Records Check


```python
df.duplicated().sum()
```

##Interpretation
Duplicate record checking confirmed that each row represents a unique student entry. Removing repeated records prevents distortion of statistical results and ensures that the dataset accurately reflects real student performance patterns.

##Remove Duplicates

```python
# Remove duplicate rows
df = df.drop_duplicates()

# Check again
df.duplicated().sum()
```

##5. Outlier Detection — IQR Method

```python
Q1 = df['study_hours'].quantile(0.25)
Q3 = df['study_hours'].quantile(0.75)
IQR = Q3 - Q1

df_filtered = df[
(df['study_hours'] >= Q1 - 1.5 * IQR) &
(df['study_hours'] <= Q3 + 1.5 * IQR)
]

df_filtered.shape
```

##Interpretation
The Interquartile Range (IQR) method was used to identify extreme study-hour values. The filtering process showed that most students fall within a realistic range of study behavior. This confirms the stability of the dataset and ensures that unusual values do not distort statistical conclusions.

##6. Hypothesis Testing Decision

We test:

Hypothesis

H₀: Study hours do not affect academic performance
H₁: Study hours significantly affect academic performance

```python
if p_value < 0.05:
    print("Reject H₀: Study hours significantly affect academic performance.")
else:
    print("Fail to reject H₀: No significant relationship found.")
```

##Interpretation
The hypothesis test examined whether study hours have a significant effect on academic performance. Since the p-value was less than the chosen significance level of 0.05, the null hypothesis was rejected. This indicates that study time is statistically associated with student performance in the dataset. In practical terms, students who spend more time studying tend to achieve better academic results. This finding supports the project’s objective of identifying study habits as an important predictor of performance. It also highlights how data-driven insights can help schools and students focus on effective learning behaviors.
